### Modelling the all records vs unique species relationship (from EDA it looks to be a logarithmic relationship as would be expected in ecological sampling patterns)

```python
import numpy as np
import statsmodels.api as sm
```

```python
model_data = annual_summary[
    (annual_summary["total_records"] > 0) &
    (annual_summary["unique_species"] > 0)
].copy()
```

#### Linear Model

##### Unique Species = Total Records + b

```python
X_linear = sm.add_constant(model_data["total_records"])

linear_model = sm.OLS(
    model_data["unique_species"],
    X_linear
).fit()

print(linear_model.summary())
```

#### Logarithmic Model

##### UniqueSpecies = log(TotalRecords) + b

```python
model_data["log_records"] = np.log(model_data["total_records"])

X_log = sm.add_constant(model_data["log_records"])

log_model = sm.OLS(
    model_data["unique_species"],
    X_log
).fit()

print(log_model.summary())
```

```python
print("Linear R²:", linear_model.rsquared)
print("Logarithmic R²:", log_model.rsquared)

print("Linear AIC:", linear_model.aic)
print("Logarithmic AIC:", log_model.aic)
```

```python
x_values = np.linspace(
    model_data["total_records"].min(),
    model_data["total_records"].max(),
    300
)

linear_predictions = linear_model.predict(
    sm.add_constant(x_values)
)

log_predictions = log_model.predict(
    sm.add_constant(np.log(x_values))
)
```

```python
plt.figure(figsize=(10, 7))

plt.scatter(
    model_data["total_records"],
    model_data["unique_species"],
    alpha=0.7,
    label="Observed annual values"
)

plt.plot(
    x_values,
    linear_predictions,
    label="Linear model"
)

plt.plot(
    x_values,
    log_predictions,
    label="Logarithmic model"
)

plt.title("Recording Effort vs Observed Species Richness")
plt.xlabel("Total records per year")
plt.ylabel("Unique species per year")
plt.legend()

plt.show()
```

### Results suggest there is very little difference between both models, linear model has slightly larger R2 and lower AIC. But they perfrom comparatively.
### Despite seeing the evidence of a plateu visually, there just is not enough data where there is a large sampling record.
### At this stage - Yes there is a relationship - but it's exact relationship is difficult to pinpoint (somewhere between linear and logarithmic)

### We can potentially move this to a spatial analysis, by creating grids to assess whether --- 'Does recorded species richness increase with recording effort (Yes), and are some areas of Kent more species-rich than expected given the amount of recording effort?, Does it hold spatially?...'

##### Annual recording effort was strongly positively associated with observed species richness (linear model R² = 0.824). A logarithmic model provided no improvement over the linear model, suggesting that the available temporal data did not provide sufficient evidence to distinguish clearly between linear and diminishing-returns relationships.
##### That is actually an important finding because it tells us that raw species counts cannot be directly interpreted as biodiversity without accounting for sampling effort - and that is precisely why the spatial analysis is needed. We cannot simply map raw species counts and call the result biodiversity.

### Spatial Analysis - After accounting for recording effort, which areas have unusually high or low observed species richness? - Maybe we could say at the end of this --- Observed biodiversity is a product of both underlying biodiversity and observation effort.

##### Potentially design 3 maps to start  1) Raw recording effort 2) Raw observed species richness 3) Species richness relative to expected richness given recording effort
