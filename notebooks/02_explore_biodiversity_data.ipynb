### Data Cleaning

#### We shall exclude the 1 Living Specimen - irrelevant to our dataset and extrememly small sample.
#### Shall we include Human Observation and preserved specimen?
##### Human observation makes up 86% of the dataset - Including Preserved Specimen may add a sample collection spatial bias as they were collected under different circumstances maybe? I think they should both be used - but in separate analysis and compared. Does preserved specimin data collection differ from human observation spatially?

```python
biodiversity["Basis of Record"].value_counts(normalize = True)
```

#### Selecting the right timeframe - data is not recorded at the same rate, this can have a significant bias on species population. E.g. Is there actually a smaller population in this year or was there just less observations.

##### Pre 1970 recording was very low - with a small peak around 1960 - most data collected around 1980-2013. Small peaks around 2019 and 2021. 1973-2012 shows consistent recording in the 1000s - this gives us a 'relatively' stable recording era.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 6))

plt.plot(year_counts.index, year_counts.values)

plt.title("Biodiversity Records by Year")
plt.xlabel("Year")
plt.ylabel("Number of records")

plt.axvline(1973, linestyle="--", label="1973")
plt.axvline(2012, linestyle="--", label="2012")
```

#### Unique species by year vs recording effort - is there a trend? - Records vs biodiversity coverage can differ

```python
species_by_year = biodiversity.groupby('Year')['Scientific name'].nunique()
```
